```python
def harmonizeLandsatC2L2_L89(img):
    img = maskLandsatC2L2(img)

    hasB10 = img.bandNames().contains('ST_B10')  # L8/L9
    oliSR = ['SR_B2','SR_B3','SR_B4','SR_B5','SR_B6','SR_B7']
    outSR = ['blue','green','red','nir','swir1','swir2']

    # SR dos sensores harmonizada
    srSelected = img.select(oliSR).multiply(0.0000275).add(-0.2).rename(outSR)
    srSelected = ee.Image(srSelected)

    # ST (Kelvin) a partir do produto de temperatura de superfície do C2 L2
    stKelvin = img.select('ST_B10').multiply(0.00341802).add(149.0).rename('st')
    srSelected = srSelected.addBands(stKelvin).toFloat()
    # srSelected = addIndices(srSelected)  # L89 não é usado abaixo, mantido como no seu script

    return srSelected.copyProperties(img, img.propertyNames())

```

```python
def maskLandsatC2L2(img):
    qa = img.select('QA_PIXEL')
    notDilated = qa.bitwiseAnd(1 << 1).eq(0)
    notCloud = qa.bitwiseAnd(1 << 3).eq(0)
    notShadow = qa.bitwiseAnd(1 << 4).eq(0)
    notSnow = qa.bitwiseAnd(1 << 5).eq(0)
    isClear = qa.bitwiseAnd(1 << 6).neq(0)

    hasRadSat = img.bandNames().contains('QA_RADSAT')
    notSaturated = ee.Image(ee.Algorithms.If(
        hasRadSat, img.select('QA_RADSAT').eq(0), ee.Image(1)
    ))

    # Correção para utilizar ee.Image.And() corretamente
    mask = ee.Image(notDilated).And(notCloud).And(notShadow).And(notSnow).And(isClear).And(notSaturated)
    return img.updateMask(mask)

```

```python
def harmonizeLandsatC2L2_l57(img):
    img = maskLandsatC2L2(img)
    hasB06 = img.bandNames().contains('ST_B6')  # L5/L7

    tmSR = ['SR_B1','SR_B2','SR_B3','SR_B4','SR_B5','SR_B7']
    outSR = ['blue','green','red','nir','swir1','swir2']

    # SR dos sensores harmonizada
    srSelected = img.select(tmSR).multiply(0.0000275).add(-0.2).rename(outSR)

    # ST (Kelvin) a partir do produto de temperatura de superfície do C2 L2
    stKelvin = img.select('ST_B6').multiply(0.00341802).add(149.0).rename('st')

    srSelected = srSelected.addBands(stKelvin).toFloat()
    srSelected = addIndices(srSelected)

    return srSelected.copyProperties(img, img.propertyNames())

```

```python
# =================== ÍNDICES (NDBSI + NDVI + LST + Wetness) ===================
def addIndices(img):
    # NDBSI
    ndbsi = img.expression(
        '((swir1 + red) - (nir + blue)) / ((swir1 + red) + (nir + blue))',
        {
            'swir1': img.select('swir1'),
            'red': img.select('red'),
            'nir': img.select('nir'),
            'blue': img.select('blue')
        }
    ).rename('NDBSI')

    # NDVI
    ndvi = img.normalizedDifference(['nir', 'red']).rename('NDVI')

    # LST em °C
    lstC = img.select('st').subtract(273.15).rename('LST')

    # Tasseled Cap (Wetness)
    X = img.select(['blue', 'green', 'red', 'nir', 'swir1', 'swir2']).toArray().toArray(1)

    TCT_TM = ee.Array([
        [0.3037, 0.2793, 0.4743, 0.5585, 0.5082, 0.1863],  # Brightness (TCB)
        [-0.2848, -0.2435, -0.5436, 0.7243, 0.0840, -0.1800],  # Greenness (TCG)
        [0.1509, 0.1973, 0.3279, 0.3406, -0.7112, -0.4572]   # Wetness (TCW)
    ])

    TCT_OLI = ee.Array([
        [0.3029, 0.2786, 0.4733, 0.5599, 0.5080, 0.1872],
        [-0.2941, -0.2430, -0.5424, 0.7276, 0.0713, -0.1608],
        [0.1511, 0.1973, 0.3283, 0.3407, -0.7117, -0.4559]
    ])

    hasB10 = img.bandNames().contains('ST_B10')  # OLI (L8/L9)
    A = ee.Array(ee.Algorithms.If(hasB10, TCT_OLI, TCT_TM))

    tcap = ee.Image(A).matrixMultiply(X).arrayProject([0]).arrayFlatten([['TCB', 'TCG', 'TCW']])

    # Banda "wetness" explicitamente exposta
    wet = tcap.select('TCW').rename('wetness')

    return img.addBands([ndbsi, ndvi, lstC, tcap, wet])

```

```python
# =================== PCA (apenas ndbsi, ndvi, lst, wet/wetness) ===================
def get_new_band_names(prefix):
    return [f'{prefix}{i}' for i in range(1, 5)]

def get_principal_components(centered, scale, region):
    arrays = centered.toArray()
    covar = arrays.reduceRegion(
        reducer=ee.Reducer.centeredCovariance(),
        geometry=region,
        scale=scale,
        maxPixels=1e9,
        tileScale=4,
        bestEffort=True
    )
    # Se covariância vier vazia, usar matriz identidade 4x4
    covar_dict = ee.Dictionary(covar)
    covar_array = ee.Array(
        ee.Algorithms.If(
            covar_dict.contains('array'),
            covar_dict.get('array'),
            ee.Array.identity(4)
        )
    )

    eigens = covar_array.eigen()
    eigen_values = eigens.slice(1, 0, 1)   # (4x1)
    eigen_vectors = eigens.slice(1, 1)     # (4x4)

    array_image = arrays.toArray(1)        # (4 x 1 x pixels)
    principal_components = ee.Image(eigen_vectors).matrixMultiply(array_image)

    sd_image = (ee.Image(eigen_values.sqrt())
                .arrayProject([0])
                .arrayFlatten([get_new_band_names('sd')]))

    return (principal_components.arrayProject([0])
            .arrayFlatten([get_new_band_names('pc')])
            .divide(sd_image))

def addPCA_on_indices(img, region_geom, scale):
    # Seleção robusta com regex (case-insensitive) e wet/wetness
    sel_any = img.select([
        '(?i)^ndbsi$',
        '(?i)^ndvi$',
        '(?i)^lst$',
        '(?i)^(wet|wetness)$'
    ])
    sel = sel_any.rename(['ndbsi','ndvi','lst','wet']).toFloat()

    # Preenche pixels mascarados com 0 para garantir médias e covariância válidas
    sel_filled = sel.unmask(0)

    # Médias com defaults seguros
    means = sel_filled.reduceRegion(
        reducer=ee.Reducer.mean(),
        geometry=region_geom,
        scale=scale,
        maxPixels=1e9,
        tileScale=4,
        bestEffort=True
    )
    d = ee.Dictionary(means)
    mean_list = ee.List(['ndbsi','ndvi','lst','wet']).map(lambda k: ee.Number(d.get(k, 0)))
    mean_img = ee.Image.constant(mean_list).rename(['ndbsi','ndvi','lst','wet'])

    # Centraliza usando a versão preenchida
    centered = sel_filled.subtract(mean_img)

    # PCs normalizados
    pcs = get_principal_components(centered, scale, region_geom)
    return img.addBands(pcs)
```

```python
# =================== PROCESSAMENTO E EXPORTAÇÃO ===================
lst_codSHP = allGeometries.reduceColumns(ee.Reducer.toList(), ['CD_MUN']).get('list')
lst_codSHP_info = lst_codSHP.getInfo()
print(f"we process {len(lst_codSHP_info)}")

for codMun in lst_codSHP_info[7:12]:  # remova o slice para processar todos primeira iteração 0:2 / 3:6 / 7:12 /
    print("procecing cidade >> ", codMun)
    featCol = allGeometries.filter(ee.Filter.eq('CD_MUN', codMun))
    feat = featCol.first()
    mask_feat = featCol.reduceToImage(['id_cod'], ee.Reducer.first()).gt(0)

    landsatAll_regL5 = (
        ee.ImageCollection('LANDSAT/LT05/C02/T1_L2')
        #.filterDate('1985-01-01', '2012-01-01')
        .filterDate('2002-03-11', '2012-01-01')
        .filterBounds(feat.geometry())
        .map(harmonizeLandsatC2L2_l57)
    )

    lst_idcod = landsatAll_regL5.reduceColumns(ee.Reducer.toList(), ['system:index']).get('list').getInfo()

    for idSystem in lst_idcod[:]:
        img_temp = ee.Image(landsatAll_regL5.filter(ee.Filter.eq('system:index', idSystem)).first())

        # --- PCA sobre NDBSI/NDVI/LST e wet/wetness (com unmask e fallback) ---
        img_temp = addPCA_on_indices(img_temp, featCol.geometry(), exportScale)

        # Recorte e propriedades (sem definir system:footprint)
        img_temp = img_temp.updateMask(mask_feat).copyProperties(feat, ['NM_MUN', 'CD_MUN'])
        img_temp = ee.Image(img_temp)  # garante tipo Image

        is_asset_exp = assetPrefix + '/' + idSystem
        print(f" exporting image to asset << {is_asset_exp} >> ")

        task = ee.batch.Export.image.toAsset(
            image=img_temp,
            description=idSystem,
            assetId=is_asset_exp,
            region=featCol.geometry().getInfo()['coordinates'],
            pyramidingPolicy={".default": "mode"},
            scale=exportScale,
            maxPixels=1e13
        )
        task.start()
        print(f" exporting image << {idSystem} >> to asset")
```
